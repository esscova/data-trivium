### 3.2 Cálculos Básicos

````python
# Crie uma nova coluna chamada `Consumo_por_Morador` que calcule o consumo de energia por morador (`Consumo_kWh` / `Número_Moradores`).
df['Consumo_por_Morador'] = df['Consumo_kWh']/df['Número_Moradores']
df.head()
````

````python
#  Calcule o consumo total de energia em todos os meses.
f"Consumo total (kWh): {df['Consumo_kWh'].sum()}"
````

### 3.3 Análise por Agrupamento

````python
# Agrupe os dados por `Tipo_Residência` e calcule a média de `Consumo_kWh` e `Consumo_por_Morador`
df.groupby('Tipo_Residência')[['Consumo_kWh', 'Consumo_por_Morador']].mean()
````

````python
# Agrupe os dados por `Mês` e calcule o consumo total de energia em cada mês.
df['Mês'] = pd.to_datetime(df['Mês'],  format='%Y-%m')
df.groupby(df['Mês'].dt.to_period('M'))['Consumo_kWh'].sum().reset_index()
````
