### 3.2 Cálculos Básicos
- Adicione uma nova coluna chamada Receita que calcula o valor total de cada venda (Preço * Quantidade_Vendida).
- Calcule a receita total de todas as vendas.

````python
# add atributo receita
df['Receita'] = df['Preço']*df['Quantidade_Vendida']
df.head(3)
````

````python
# receita total
print(df.Receita.sum())
````

### 3.3 Análise por Agrupamento
- Agrupe os dados por Produto e calcule a média de Preço e a soma de Quantidade_Vendida para cada produto.
- Agrupe os dados por Região e calcule a receita total por região.

````python
# agrupados por produto com média de preço e soma da quantidade vendida
(
    df.groupby('Produto')
     .agg({'Preço':'mean', 'Quantidade_Vendida':'sum'})
     .rename(columns={'Preço': 'Preço_Médio', 'Quantidade_Vendida': 'Total_Vendido'})
)
````

````python
# agrupados por regiao com receita total
df.groupby('Região')['Receita'].sum()
````

### 3.4 Filtros
- Filtre as vendas onde a Quantidade_Vendida é maior que 10.
- Filtre as vendas realizadas na região Sul com preço superior a 1000.

````python
# vendas  = quantidade vendida > 10
df[df['Quantidade_Vendida']>10]
````

````python
# vendas regiao sul = preco > 1000
df[(df['Região']=='Sul') & (df['Preço']>1000)]
#df.query("Região == 'Sul' and Preço > 1000")
````

### 3.5 Extra (Desafio)
- Converta a coluna Data_Venda para o formato datetime.
- Crie uma nova coluna que indique o dia da semana de cada venda.
- Identifique qual dia da semana teve a maior receita total.

````python
# converter atributo para formato date
df['Data_Venda']=pd.to_datetime(df['Data_Venda'])
df.dtypes
````

````python
# atributo dia da semana
df['Dia_Semana'] = df['Data_Venda'].dt.day_name()
df.head()
````

````python
# dia da semanha com maior receita
df.groupby('Dia_Semana')['Receita'].sum().idxmax()
````
