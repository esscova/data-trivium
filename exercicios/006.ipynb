## Solução

### 1. Dependencias e configurações

````python
import sys
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import plotly
import plotly.express as px

print(f"""
=== Versões Utilizadas ===
Python: {sys.version.split()[0]}
Pandas: {pd.__version__}
Numpy: {np.__version__}
Matplotlib: {matplotlib.__version__}
Plotly: {plotly.__version__}
""")
````

### 2. Coleta de dados

````python
try:
    df = pd.DataFrame(
        {
            'Pedido_ID': [101, 102, 103, 104, 105, 106, 107, 108],
            'Item': ['Pizza', 'Hambúrguer', 'Salada', 'Pizza', 'Hambúrguer', 'Salada', 'Pizza', 'Hambúrguer'],
            'Categoria': ['Prato Principal', 'Prato Principal', 'Entrada', 'Prato Principal', 'Prato Principal', 'Entrada', 'Prato Principal', 'Prato Principal'],
            'Preço': [45.0, 25.0, 15.0, 50.0, 28.0, 12.0, 48.0, 30.0],
            'Quantidade': [2, 3, 1, 1, 4, 2, 3, 2],
            'Data_Pedido': ['2023-06-01', '2023-06-01', '2023-06-02', '2023-06-02', '2023-06-03', '2023-06-03', '2023-06-04', '2023-06-04'],
            'Turno': ['Jantar', 'Almoço', 'Jantar', 'Almoço', 'Jantar', 'Almoço', 'Jantar', 'Almoço']
        }
    )
    print("\n ===Dados Carregados ===")
except Exception as e:
        raise ValueError(f"Erro ao carregar dados:\n{e}")
````

## Tarefas

### 1. Exploração Inicial

````python
# Exiba as primeiras 3 linhas do DataFrame.
df.head(3)
````

````python
# Verifique se há valores ausentes no conjunto de dados.
df.isna().sum()
````

````python
# Mostre um resumo estatístico das colunas `Preço` e `Quantidade`.
df.drop("Pedido_ID", axis=1).describe()
````

````python
# **Visualização**: Crie um gráfico de dispersão com Plotly mostrando `Preço` (eixo x) versus `Quantidade` (eixo y), com pontos coloridos por `Item`.
fig = px.scatter(
    df,
    x='Preço',
    y='Quantidade',
    color='Item',
    title='Preço vs Quantidade',
)
fig.show()
````

### 2. Cálculos Básicos

````python
df2 = df.copy()
````

````python
#- Crie uma nova coluna chamada `Total_Pedido` que calcule o valor total de cada pedido (`Preço` * `Quantidade`).
df2['Total_Pedido'] = df2['Preço']*df2['Quantidade']
df2.sample()
````

````python
#- Calcule o faturamento total do restaurante (soma de `Total_Pedido`).
f'Total (R$): {df2['Total_Pedido'].sum()}'
````

````python
#- **Visualização**: Crie um gráfico de barras com Plotly mostrando o `Total_Pedido` para cada `Pedido_ID`.
fig = px.bar(
    df2, y='Total_Pedido', x='Pedido_ID',
    title='Total (R$) por Pedido',
    text='Total_Pedido'
)
fig.update_xaxes(tickmode='linear', dtick=1) # todos xaxex
fig.add_hline( # plot da media de vendas
    y=df2['Total_Pedido'].mean(),
    line_dash='dash',
    line_color='red',
    annotation_text=f"Média: R$ {df2['Total_Pedido'].mean():.2f}",
)
fig.show()
````
