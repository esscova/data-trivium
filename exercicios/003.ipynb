#### 3.1.3 Mostre as estatísticas descritivas (média, mínimo, máximo, etc.) das colunas de notas (Matemática, Ciências, História).

````python
# estatisticas descritivas das notas
#df.iloc[:, :-1].describe()
df.drop(df.columns[-1], axis=1).describe()
````

### 3.2. Cálculos Básicos

#### 3.2.1 Crie uma nova coluna chamada Média_Geral que calcule a média das notas das três disciplinas para cada aluno.

````python
# media
df['Média_Geral'] = (df["Matemática"]+df["Ciências"]+df["História"])/3
df.head()
````

#### 3.2.2 Identifique o aluno com a maior média geral.

````python
df.iloc[df['Média_Geral'].idxmax()]
````

### 3.3 Análise por Agrupamento

#### 3.3.1 Agrupe os dados por Turma e calcule a média das notas de cada disciplina e da Frequência (%).

````python
df.groupby('Turma')[["Matemática", "Ciências", "História", "Frequência (%)"]].mean()
````

#### 3.3.2 Crie uma nova coluna Aprovado que indique se o aluno foi aprovado (média geral ≥ 7.0) ou reprovado (média geral < 7.0). Em seguida, conte quantos alunos foram aprovados em cada turma.

````python
df['Aprovado'] = np.where(df['Média_Geral']<7,0,1)
df.head()
````

````python
df.groupby('Turma')['Aprovado'].sum().reset_index()
````

### 3.4 Filtros

#### 3.4.1 Filtre os alunos que têm frequência superior a 80% e média geral maior ou igual a 7.0.

````python
df[(df['Frequência (%)']>80) & (df['Média_Geral']>=70)]
````

#### 3.4.2 Filtre os alunos que obtiveram nota inferior a 6.0 em pelo menos uma das disciplinas.

````python
df[df[df["História Matemática Ciências".split()]<6.0].any(axis=1)]
````
