# Introdução ao Primeiro Neurônio Artificial

## Bloco 1: Teoria e Contexto

O Perceptron é um dos primeiros algoritmos de Machine Learning já criados.

Ele foi proposto por Frank Rosenblatt em 1958 e é considerado a base das redes neurais modernas e foi inspirado no neurônio biológico. Ele recebe informações (entradas), processa essas informações (soma ponderada) e toma uma decisão (função de ativação), gerando uma saída.

Imagine que queremos classificar:

| Altura | Classe |
| ------ | ------ |
| 1.50   | Baixa  |
| 1.90   | Alta   |


O Perceptron tentará aprender uma regra matemática para decidir automaticamente a qual classe um novo valor pertence.

## Bloco 2: Definição Formal

O algoritmo Perceptron é um tipo de algoritmo de **aprendizado supervisionado** utilizado para problemas de **classificação binária**. Ele é a base para redes neurais mais complexas e é utilizado para determinar se uma entrada pertence a uma classe ou não.

### Estrutura Matemática

Matematicamente, o Perceptron pode ser representado da seguinte forma:

$$ y = f\left(\sum_{i=1}^{n} w_i \cdot x_i + b\right) $$

Onde:
- **$y$** → saída do Perceptron
- **$w_i$** → pesos associados a cada entrada $x_i$
- **$b$** → bias (viés), um valor constante que ajuda a ajustar a função de ativação
- **$f$** → função de ativação, geralmente a **função Degrau**, que retorna 1 se a soma ponderada for maior ou igual a zero, e 0 caso contrário

## Bloco 3: Anatomia Matemática do Perceptron

Até aqui vimos a fórmula geral utilizada pelo Perceptron. Agora vamos analisar cada etapa do cálculo para entender como o algoritmo toma decisões.

### 2.1 Soma Ponderada

O primeiro passo realizado pelo Perceptron é calcular uma combinação linear das entradas.

Essa operação recebe o nome de **soma ponderada**, pois cada entrada possui um peso que representa sua importância para a decisão final.

A fórmula é:

$$
z = \sum_{i=1}^{n} w_i x_i + b
$$

Onde:

* $x_i$ representa uma entrada
* $w_i$ representa o peso associado à entrada
* $b$ representa o bias (viés)
* $z$ representa o resultado da soma ponderada

O valor de $z$ ainda não é uma classificação. Ele é apenas uma pontuação que será utilizada pela função de ativação para tomar uma decisão.

#### Exemplo Teórico

Suponha que temos duas entradas:

$$
x_1 = 2
$$

$$
x_2 = 3
$$

Com os seguintes pesos:

$$
w_1 = 0.5
$$

$$
w_2 = 0.2
$$

E um bias igual a:

$$
b = 1
$$

Substituindo os valores na fórmula:

$$
z = (2 \times 0.5) + (3 \times 0.2) + 1
$$

$$
z = 1 + 0.6 + 1
$$

$$
z = 2.6
$$

O resultado da soma ponderada é:

$$
z = 2.6
$$

#### Implementação em Python

```python
x1 = 2
x2 = 3

w1 = 0.5
w2 = 0.2

bias = 1

z = (x1 * w1) + (x2 * w2) + bias

print(f"Soma ponderada: {z}")
```

### 2.2 Função Degrau

Após calcular a soma ponderada, o Perceptron precisa decidir a qual classe a observação pertence.

Para isso utilizamos uma função de ativação. A mais simples e tradicional é a **Função Degrau (Step Function)**.

Sua definição é:

$$
f(z)=
\begin{cases}
1, & z \geq 0 \
0, & z < 0
\end{cases}
$$

Isso significa que:

* Se o resultado da soma ponderada for maior ou igual a zero, a saída será 1.
* Se o resultado for menor que zero, a saída será 0.

#### Exemplo Teórico

Considere:

$$
z = 2.6
$$

Como:

$$
2.6 \geq 0
$$

Temos:

$$
f(2.6) = 1
$$

Agora considere:

$$
z = -1.4
$$

Como:

$$
-1.4 < 0
$$

Temos:

$$
f(-1.4) = 0
$$

#### Implementação em Python

```python
def funcao_degrau(z):
    if z >= 0:
        return 1
    else:
        return 0

#testando a função
print(funcao_degrau(2.6))
print(funcao_degrau(-1.4))
```

### Conectando as Etapas

O funcionamento básico do Perceptron pode ser resumido em duas etapas:

1. Calcular a soma ponderada.
2. Aplicar a função degrau para obter a classificação.

Visualmente:

$$
Entradas \rightarrow Soma\ Ponderada \rightarrow Função\ Degrau \rightarrow Saída
$$

Essa sequência representa o núcleo do funcionamento de um neurônio artificial simples.
