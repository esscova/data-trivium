# Introdução ao Primeiro Neurônio Artificial

## Bloco 1: Teoria e Contexto

O Perceptron é um dos primeiros algoritmos de Machine Learning já criados.

Ele foi proposto por Frank Rosenblatt em 1958 e é considerado a base das redes neurais modernas e foi inspirado no neurônio biológico. Ele recebe informações (entradas), processa essas informações (soma ponderada) e toma uma decisão (função de ativação), gerando uma saída.

Imagine que queremos classificar:

| Altura | Classe |
| ------ | ------ |
| 1.50   | Baixa  |
| 1.90   | Alta   |


O Perceptron tentará aprender uma regra matemática para decidir automaticamente a qual classe um novo valor pertence.

## Bloco 2: Definição Formal

O algoritmo Perceptron é um tipo de algoritmo de **aprendizado supervisionado** utilizado para problemas de **classificação binária**. Ele é a base para redes neurais mais complexas e é utilizado para determinar se uma entrada pertence a uma classe ou não.

### Estrutura Matemática

Matematicamente, o Perceptron pode ser representado da seguinte forma:

$$ y = f\left(\sum_{i=1}^{n} w_i \cdot x_i + b\right) $$

Onde:
- **$y$** → saída do Perceptron
- **$w_i$** → pesos associados a cada entrada $x_i$
- **$b$** → bias (viés), um valor constante que ajuda a ajustar a função de ativação
- **$f$** → função de ativação, geralmente a **função Degrau**, que retorna 1 se a soma ponderada for maior ou igual a zero, e 0 caso contrário

## Bloco 3: Primeira Previsão do Perceptron

Agora que já entendemos como a soma ponderada é calculada e como a função degrau transforma esse valor em uma decisão, podemos unir essas duas etapas para realizar nossa primeira previsão.

---

### Como o Perceptron Faz uma Previsão?

O processo ocorre em duas etapas:

#### Etapa 1 - Calcular a Soma Ponderada

$$
z = \sum_{i=1}^{n} w_i x_i + b
$$

O resultado é uma pontuação numérica.

#### Etapa 2 - Aplicar a Função Degrau

$$
f(z)=
\begin{cases}
1, & z \geq 0 \
0, & z < 0
\end{cases}
$$

O resultado será uma classificação binária:

* Classe 0
* Classe 1

---

### Exemplo Teórico

Considere as seguintes entradas:

| Variável | Valor |
| -------- | ----- |
| $x_1$    | 2     |
| $x_2$    | 3     |

Pesos:

| Peso  | Valor |
| ----- | ----- |
| $w_1$ | 0.5   |
| $w_2$ | 0.2   |

Bias:

$$
b = 1
$$

#### Calculando a Soma Ponderada

$$
z = (2 \times 0.5) + (3 \times 0.2) + 1
$$

$$
z = 2.6
$$

#### Aplicando a Função Degrau

Como:

$$
2.6 \geq 0
$$

Temos:

$$
f(2.6) = 1
$$

Portanto, a previsão do Perceptron é:

$$
y = 1
$$

### Implementação Passo a Passo
#### Criando a Função Degrau

`````python
def funcao_degrau(z):
    if z >= 0:
        return 1
    return 0
`````

#### Calculando a Soma Ponderada

`````python
x1 = 2
x2 = 3

w1 = 0.5
w2 = 0.2

bias = 1

z = (x1 * w1) + (x2 * w2) + bias

print(z)
`````

#### Realizando a Classificação

`````python
previsao = funcao_degrau(z)

print(previsao)
`````

#### Criando uma Função de Previsão

Podemos encapsular todo o processo em uma única função.

`````python
def predict(x1, x2, w1, w2, bias):

    z = (x1 * w1) + (x2 * w2) + bias

    if z >= 0:
        return 1

    return 0
`````

Utilizando a função:

`````python
resultado = predict(
    x1=2,
    x2=3,
    w1=0.5,
    w2=0.2,
    bias=1
)

print(resultado)
`````

### Generalizando para Múltiplas Entradas

Na prática, problemas reais possuem diversas características de entrada.

Por exemplo:

| Altura | Peso | Idade | Classe |
| ------ | ---- | ----- | ------ |
| 1.75   | 80   | 30    | 1      |
| 1.60   | 55   | 20    | 0      |

Nesses casos utilizamos vetores para armazenar entradas e pesos.

`````python
import numpy as np

X = np.array([2, 3])

W = np.array([0.5, 0.2])

bias = 1

z = np.dot(X, W) + bias

previsao = funcao_degrau(z)

print(previsao)
`````

### Limitação Importante

Observe que até este momento os pesos foram definidos manualmente.

```python
w1 = 0.5
w2 = 0.2
```

Mas em Machine Learning não queremos escolher pesos manualmente.

O objetivo é que o algoritmo descubra automaticamente os melhores valores observando exemplos rotulados.

Esse processo recebe o nome de treinamento e será o próximo passo do nosso estudo.

## Bloco 4: O Processo de Aprendizado

No bloco anterior, utilizamos pesos definidos manualmente para realizar previsões. No entanto, em problemas reais não sabemos quais são os melhores pesos para cada entrada.

É justamente nesse ponto que surge o principal diferencial do Perceptron: sua capacidade de **aprender a partir dos próprios erros**.

O aprendizado ocorre por meio de um processo iterativo de tentativa e erro. A cada exemplo apresentado durante o treinamento, o algoritmo verifica se sua previsão está correta. Caso exista erro, os pesos e o viés são ajustados para aumentar as chances de acerto nas próximas previsões.

Em outras palavras, o Perceptron aprende corrigindo gradualmente seus próprios erros.

### Como o Processo Funciona?

Para cada exemplo do conjunto de treinamento, o algoritmo executa os seguintes passos:

1. Recebe os valores de entrada.
2. Calcula a soma ponderada.
3. Aplica a função de ativação e gera uma previsão ($\hat{y}$).
4. Compara a previsão com o valor esperado ($y$).
5. Calcula o erro.
6. Atualiza os pesos e o viés quando necessário.

Representando esse fluxo:

```text
Entradas
    ↓
Soma Ponderada
    ↓
Função Degrau
    ↓
Previsão (ŷ)
    ↓
Comparação com y
    ↓
Cálculo do Erro
    ↓
Atualização dos Pesos
```

---

### Cálculo do Erro

Após realizar uma previsão, o Perceptron compara o resultado obtido com a resposta correta.

O erro é calculado pela seguinte equação:

$$
Erro = y - \hat{y}
$$

Onde:

* $y$ é o valor real (resposta correta)
* $\hat{y}$ é o valor previsto pelo modelo

#### Possíveis Resultados

| Valor Real ($y$) | Previsão ($\hat{y}$) | Erro |
| ---------------- | -------------------- | ---- |
| 1                | 1                    | 0    |
| 0                | 0                    | 0    |
| 1                | 0                    | 1    |
| 0                | 1                    | -1   |

Observe que:

* Erro igual a 0 significa que o modelo acertou.
* Erro igual a 1 significa que o modelo deveria ter previsto 1, mas previu 0.
* Erro igual a -1 significa que o modelo deveria ter previsto 0, mas previu 1.

---

### Atualização dos Pesos

Quando ocorre um erro, os pesos são ajustados utilizando a seguinte regra:

$$
w_i = w_i + \eta \cdot Erro \cdot x_i
$$

Onde:

* $w_i$ é o peso atual
* $\eta$ (eta) é a taxa de aprendizado (*learning rate*)
* $Erro$ é a diferença entre o valor real e o previsto
* $x_i$ é a entrada associada ao peso

O viés também é atualizado:

$$
b = b + \eta \cdot Erro
$$

---

### Intuição da Regra de Atualização

A regra de atualização faz com que o Perceptron mova sua fronteira de decisão na direção correta.

Podemos interpretar os resultados da seguinte forma:

| Erro | Significado                    | Ação                 |
| ---- | ------------------------------ | -------------------- |
| 0    | Acertou                        | Nenhuma alteração    |
| 1    | Deveria prever 1, mas previu 0 | Aumenta pesos e viés |
| -1   | Deveria prever 0, mas previu 1 | Diminui pesos e viés |

Em resumo:

* Quando o Perceptron acerta, ele mantém o conhecimento atual.
* Quando erra para menos, ele fortalece a influência das entradas.
* Quando erra para mais, ele reduz a influência das entradas.

Esse mecanismo simples de correção é o que permite ao algoritmo aprender padrões presentes nos dados.

## Bloco 6: Transformando o Processo de Aprendizado em Código

No exemplo anterior, calculamos manualmente:

1. A soma ponderada.
2. A previsão.
3. O erro.
4. A atualização dos pesos.
5. A atualização do viés.

Embora isso seja útil para compreender o funcionamento do algoritmo, seria inviável repetir esse processo manualmente para centenas ou milhares de exemplos.

Por esse motivo, implementamos um método chamado `fit()`, responsável por automatizar todo o processo de treinamento.

Em Machine Learning, o método `fit()` é tradicionalmente utilizado para ajustar os parâmetros de um modelo aos dados de treinamento.

### Nossa Primeira Versão do Método fit()
Vamos começar treinando o Perceptron com apenas um único exemplo.

`````python
import numpy as np

def funcao_degrau(z):
    return 1 if z >= 0 else 0


def fit(x, y, learning_rate=0.1):

    #inicialização dos parâmetros
    weights = np.zeros(len(x))
    bias = -1

    print("Pesos iniciais:", weights)
    print("Bias inicial:", bias)

    #soma ponderada
    z = np.dot(x, weights) + bias

    #previsão
    y_pred = funcao_degrau(z)

    #erro
    erro = y - y_pred

    #atualização dos pesos
    weights = weights + learning_rate * erro * x

    #atualização do viés
    bias = bias + learning_rate * erro

    print("\nApós atualização:")
    print("Pesos:", weights)
    print("Bias:", bias)

    return weights, bias
`````

### Executando o Treinamento

`````python
x = np.array([2, 3])
y = 1

weights, bias = fit(x, y)
`````

Observe que obtivemos exatamente os mesmos valores calculados manualmente no exemplo anterior.

### Fazendo uma Nova Previsão

Agora podemos verificar se o modelo aprendeu algo após a atualização.

`````python
z = np.dot(x, weights) + bias

print("Nova soma ponderada:", z)
print("Nova previsão:",funcao_degrau(z) )
`````

Após apenas uma atualização, o Perceptron já passou a classificar corretamente esse exemplo.

### Limitação da Nossa Implementação

Nossa função ainda possui uma limitação importante:

* Treina apenas um exemplo.
* Realiza apenas uma atualização.
* Não percorre um conjunto completo de dados.

Em aplicações reais, precisamos:

* Treinar vários exemplos.
* Repetir o treinamento diversas vezes.
* Atualizar os pesos continuamente.

Para isso, no próximo passo criaremos um método `fit()` capaz de percorrer todo um conjunto de treinamento utilizando laços de repetição.

## Bloco 7: Evoluindo o Método `fit()`

Na implementação anterior, treinamos o Perceptron utilizando apenas um único exemplo.

Embora isso seja suficiente para compreender a mecânica de atualização dos pesos, um modelo de Machine Learning precisa aprender observando diversos exemplos.

Vamos evoluir gradualmente nosso método `fit()`.

### Versão 1: Treinando um Único Exemplo

Até agora trabalhamos com:

```python
x = np.array([2, 3])

y = 1
```

O treinamento ocorre apenas uma vez.

Visualmente:

```text
Exemplo 1
    ↓
Previsão
    ↓
Erro
    ↓
Atualização
```

Essa abordagem possui uma limitação importante: o modelo aprende apenas com uma observação.

---

### Versão 2: Treinando Múltiplos Exemplos

Agora vamos fornecer um conjunto de treinamento.

`````python
import numpy as np

X = np.array([
    [2, 3],
    [1, 1],
    [4, 5]
])

y = np.array([
    1,
    0,
    1
])
`````

Observe que:

* Cada linha de `X` representa uma observação.
* Cada elemento de `y` representa a resposta correta daquela observação.

Visualmente:

```text
[2, 3] → Classe 1

[1, 1] → Classe 0

[4, 5] → Classe 1
```

### Percorrendo os Dados

Precisamos processar cada exemplo individualmente.

Para isso utilizamos um laço de repetição.

`````python
for idx, x_i in enumerate(X):

    print("Entrada:", x_i)

    print("Classe:", y[idx])

    print()
`````

Agora conseguimos acessar cada observação e sua respectiva classe.

---

### Atualizando os Pesos para Cada Exemplo

Vamos incorporar o processo de aprendizado.

`````python
def funcao_degrau(z):
    return 1 if z >= 0 else 0


weights = np.zeros(2)

bias = -1

learning_rate = 0.1

for idx, x_i in enumerate(X):

    z = np.dot(x_i, weights) + bias

    y_pred = funcao_degrau(z)

    erro = y[idx] - y_pred

    weights = (
        weights
        + learning_rate
        * erro
        * x_i
    )

    bias = (
        bias
        + learning_rate
        * erro
    )

    print(f"Exemplo {idx + 1}")
    print("Erro:", erro)
    print("Pesos:", weights)
    print("Bias:", bias)
    print()
`````

Agora o Perceptron aprende observando vários exemplos.

---

### Uma Nova Limitação

Observe que percorremos os dados apenas uma vez.

```text
Exemplo 1
Exemplo 2
Exemplo 3
```

Após chegar ao último exemplo, o treinamento termina.

Na prática, normalmente uma única passagem pelos dados não é suficiente para que o modelo aprenda corretamente.

Precisamos apresentar os mesmos exemplos diversas vezes.

## Bloco 8: Introduzindo o Conceito de Época

Uma época (*epoch*) representa uma passagem completa por todos os dados de treinamento.

Por exemplo:

```text
Época 1
    ↓
Exemplo 1
Exemplo 2
Exemplo 3

Época 2
    ↓
Exemplo 1
Exemplo 2
Exemplo 3

Época 3
    ↓
Exemplo 1
Exemplo 2
Exemplo 3
```

A cada época os pesos são refinados.

---

### Adicionando o Loop de Épocas

Agora teremos dois laços:

* Um para as épocas.
* Um para os exemplos.

`````python
epochs = 5

for epoch in range(epochs):

    print(
        f"\n=== Época {epoch + 1} ==="
    )

    for idx, x_i in enumerate(X):

        z = np.dot(x_i, weights) + bias

        y_pred = funcao_degrau(z)

        erro = y[idx] - y_pred

        weights += (
            learning_rate
            * erro
            * x_i
        )

        bias += (
            learning_rate
            * erro
        )

    print(
        "Pesos atuais:",
        weights
    )

    print(
        "Bias atual:",
        bias
    )
`````

### Entendendo os Dois Loops

O treinamento do Perceptron pode ser resumido da seguinte forma:

```python
for epoch in range(epochs):

    for exemplo in X:

        prever()

        calcular_erro()

        atualizar_pesos()
```

O primeiro laço controla quantas vezes veremos os dados.

O segundo laço percorre cada observação do conjunto de treinamento.

Essa estrutura será utilizada na implementação final da classe Perceptron.


## Bloco 9: Construindo a Classe Perceptron

Nosso objetivo agora é reunir tudo o que construímos em uma única estrutura.

---

### O Que Nossa Classe Deve Fazer?

Queremos uma classe capaz de:

### Método Construtor (**init**)

Toda classe Python pode possuir um método especial chamado `__init__()`.

Esse método é executado automaticamente quando criamos um objeto.

Nele armazenaremos:

* Taxa de aprendizado.
* Quantidade de épocas.

`````python
class Perceptron:

    def __init__(
        self,
        learning_rate=0.1,
        epochs=100
    ):

        self.learning_rate = learning_rate

        self.epochs = epochs

        self.weights = None

        self.bias = None
`````

Neste momento o modelo existe, mas ainda não foi treinado.

### Método fit()

O método `fit()` será responsável pelo treinamento.

Dentro dele colocaremos exatamente a lógica estudada anteriormente:

1. Inicializar pesos.
2. Percorrer as épocas.
3. Percorrer os exemplos.
4. Calcular previsões.
5. Calcular erros.
6. Atualizar pesos e viés.

---

#### Estrutura Inicial

```python
class Perceptron:

    def fit(self, X, y):

        pass
```

---

#### Inicializando os Pesos

Antes de começar o treinamento precisamos criar um peso para cada característica do conjunto de dados.

```python
def fit(self, X, y):

    n_samples, n_features = X.shape

    self.weights = np.zeros(n_features)

    self.bias = 0
```

#### Entendendo o X.shape

Suponha:

`````python
X = np.array([
    [2, 3],
    [1, 1],
    [4, 5]
])
print(X.shape)
`````

Isso significa:

* 3 observações (linhas)
* 2 características (colunas)

Portanto:

```python
n_samples = 3

n_features = 2
```

Como temos duas características, precisamos de dois pesos.

#### Implementando o Treinamento

Agora adicionamos os dois laços que estudamos anteriormente.

```python
def fit(self, X, y):

    n_samples, n_features = X.shape

    self.weights = np.zeros(n_features)

    self.bias = 0

    for _ in range(self.epochs):

        for idx, x_i in enumerate(X):

            z = (
                np.dot(
                    x_i,
                    self.weights
                )
                + self.bias
            )

            y_pred = (
                1
                if z >= 0
                else 0
            )

            erro = (
                y[idx]
                - y_pred
            )

            self.weights += (
                self.learning_rate
                * erro
                * x_i
            )

            self.bias += (
                self.learning_rate
                * erro
            )
```

Observe que essa implementação é exatamente a mesma lógica que utilizamos anteriormente. A diferença é que agora ela está organizada dentro da classe.

---

### Método predict()

Após o treinamento, precisamos utilizar os pesos aprendidos para fazer novas previsões.

Para isso criaremos o método `predict()`.

```python
def predict(self, X):

    z = (
        np.dot(
            X,
            self.weights
        )
        + self.bias
    )

    return (
        1
        if z >= 0
        else 0
    )
```

Esse método recebe uma entrada, calcula a soma ponderada e aplica a função degrau.

---

### Classe Completa

Agora podemos reunir tudo em uma única implementação.

`````python
import numpy as np


class Perceptron:

    def __init__(
        self,
        learning_rate=0.1,
        epochs=100
    ):

        self.learning_rate = learning_rate

        self.epochs = epochs

        self.weights = None

        self.bias = None

    def fit(self, X, y):

        n_samples, n_features = X.shape

        self.weights = np.zeros(n_features)

        self.bias = 0

        for _ in range(self.epochs):

            for idx, x_i in enumerate(X):

                z = (
                    np.dot(
                        x_i,
                        self.weights
                    )
                    + self.bias
                )

                y_pred = (
                    1
                    if z >= 0
                    else 0
                )

                erro = (
                    y[idx]
                    - y_pred
                )

                self.weights += (
                    self.learning_rate
                    * erro
                    * x_i
                )

                self.bias += (
                    self.learning_rate
                    * erro
                )

    def predict(self, X):

        z = (
            np.dot(
                X,
                self.weights
            )
            + self.bias
        )

        return (
            1
            if z >= 0
            else 0
        )
`````

### Utilizando o Modelo

#### Criando o Conjunto de Dados

`````python
X = np.array([
    [2, 3],
    [1, 1],
    [4, 5]
])

y = np.array([
    1,
    0,
    1
])
`````

#### Treinando o Modelo

`````python
perceptron = Perceptron(
    learning_rate=0.1,
    epochs=100
)

perceptron.fit(X, y)
`````

#### Visualizando os Parâmetros Aprendidos

`````python
print("Pesos:", perceptron.weights)
print("Bias:", perceptron.bias)
`````

#### Fazendo uma Previsão

`````python
nova_amostra = np.array([3, 4])

resultado = perceptron.predict(
    nova_amostra
)

print(resultado)
`````

## Resumo do Fluxo Completo

Todo o funcionamento do Perceptron pode ser resumido da seguinte forma:

```text
Dados de Treinamento
         ↓
      fit()
         ↓
 Aprendizado dos Pesos
         ↓
     predict()
         ↓
      Previsão
````

Neste notebook construímos um Perceptron funcional do zero, utilizando conceitos fundamentais de Álgebra Linear, Programação Orientada a Objetos e Aprendizado de Máquina.

Mais importante do que utilizar uma implementação pronta foi compreender cada etapa do processo: desde a soma ponderada e a função de ativação até o mecanismo de atualização dos pesos responsável pelo aprendizado do modelo.

### Um Desafio
A implementação desenvolvida neste notebook foi projetada para realizar previsões sobre uma única amostra por vez.

Por exemplo:

```python

perceptron.predict(nova_amostra)
```

```python
novas_amostras = np.array([
```
